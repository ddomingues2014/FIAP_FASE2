Otimização dos hiperparametros utilizando Algoritmo Genético

Nesta etapa será implementado um algoritmo genético para buscar automaticamente os melhores hiperparametros dos modelos KNN, SVM e Random Forest.

Cada individuo da população representa uma combinação de hiperparametros. A qualidade de cada individuo é calculada com validação cruzada estratificada utilizando recall como função de fitness. Para este problema o recall é importante porque penaliza os casos positivos que o modelo classificou como negativos.

O processo utiliza seleção por torneio, crossover uniforme, mutação e elitismo. A semente permite reproduzir os mesmos resultados em novas execuções.


```python
#Bibliotecas utilizadas pelo algoritmo genetico e pela validacao dos modelos
import random
from copy import deepcopy

import numpy as np
from sklearn.base import clone
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier

#Semente utilizada para reproduzir a populacao inicial, cruzamentos e mutacoes
SEMENTE_AG = 42
random.seed(SEMENTE_AG)
np.random.seed(SEMENTE_AG)
```

Configuração do espaço de busca

O espaço de busca informa quais valores o algoritmo genético poderá testar para cada hiperparametro. Foram selecionados parametros compatíveis com os três classificadores já utilizados no projeto.


```python
#Espaco de busca dos hiperparametros para cada modelo
espacos_busca = {
    "KNN": {
        "n_neighbors": list(range(1, 26, 2)),
        "weights": ["uniform", "distance"],
        "p": [1, 2]
    },
    "SVM": {
        "C": [0.001, 0.01, 0.1, 1, 10, 100],
        "class_weight": [None, "balanced"],
        "max_iter": [2000, 5000, 10000]
    },
    "Random Forest": {
        "n_estimators": [100, 200, 300, 400, 500],
        "max_depth": [None, 3, 5, 8, 12, 16, 20],
        "min_samples_split": [2, 4, 6, 8, 10],
        "min_samples_leaf": [1, 2, 3, 4, 5],
        "max_features": ["sqrt", "log2", None],
        "class_weight": [None, "balanced"]
    }
}

#Fabricas dos modelos. Cada funcao recebe um dicionario de hiperparametros
def criar_knn(parametros):
    return KNeighborsClassifier(**parametros)


def criar_svm(parametros):
    return LinearSVC(random_state=SEMENTE_AG, **parametros)


def criar_random_forest(parametros):
    return RandomForestClassifier(
        random_state=SEMENTE_AG,
        n_jobs=-1,
        **parametros
    )

fabricas_modelos = {
    "KNN": criar_knn,
    "SVM": criar_svm,
    "Random Forest": criar_random_forest
}
```

Funções do Algoritmo Genético

As funções abaixo executam as etapas principais:

1. Criar a população inicial de forma aleatória.
2. Avaliar cada individuo por validação cruzada estratificada.
3. Selecionar os pais por torneio.
4. Combinar os genes dos pais utilizando crossover uniforme.
5. Aplicar mutações para manter diversidade na população.
6. Preservar os melhores individuos por elitismo.


```python
def criar_individuo(espaco_busca):
    """Cria uma combinacao aleatoria de hiperparametros."""
    return {
        parametro: random.choice(valores)
        for parametro, valores in espaco_busca.items()
    }


def criar_populacao(espaco_busca, tamanho_populacao):
    """Cria a populacao inicial do algoritmo genetico."""
    return [
        criar_individuo(espaco_busca)
        for _ in range(tamanho_populacao)
    ]


def chave_individuo(individuo):
    """Transforma o individuo em uma chave para reutilizar avaliacoes anteriores."""
    return tuple(sorted(individuo.items()))


def avaliar_individuo(individuo, fabrica_modelo, x_treino, y_treino, validacao, cache):
    """Calcula o recall medio do individuo utilizando validacao cruzada."""
    chave = chave_individuo(individuo)

    if chave not in cache:
        modelo = fabrica_modelo(individuo)
        resultados = cross_val_score(
            modelo,
            x_treino,
            y_treino,
            cv=validacao,
            scoring="recall",
            n_jobs=-1,
            error_score="raise"
        )
        cache[chave] = resultados.mean()

    return cache[chave]


def selecionar_por_torneio(populacao_avaliada, tamanho_torneio):
    """Seleciona o melhor individuo de uma amostra aleatoria da populacao."""
    competidores = random.sample(
        populacao_avaliada,
        k=min(tamanho_torneio, len(populacao_avaliada))
    )
    vencedor = max(competidores, key=lambda item: item[1])
    return deepcopy(vencedor[0])


def aplicar_crossover(pai_1, pai_2, taxa_crossover):
    """Combina os hiperparametros dos pais utilizando crossover uniforme."""
    if random.random() >= taxa_crossover:
        return deepcopy(pai_1), deepcopy(pai_2)

    filho_1 = {}
    filho_2 = {}

    for parametro in pai_1:
        if random.random() < 0.5:
            filho_1[parametro] = pai_1[parametro]
            filho_2[parametro] = pai_2[parametro]
        else:
            filho_1[parametro] = pai_2[parametro]
            filho_2[parametro] = pai_1[parametro]

    return filho_1, filho_2


def aplicar_mutacao(individuo, espaco_busca, taxa_mutacao):
    """Altera genes do individuo de acordo com a taxa de mutacao."""
    individuo_mutado = deepcopy(individuo)

    for parametro, valores in espaco_busca.items():
        if random.random() < taxa_mutacao:
            individuo_mutado[parametro] = random.choice(valores)

    return individuo_mutado


def executar_algoritmo_genetico(
    fabrica_modelo,
    espaco_busca,
    x_treino,
    y_treino,
    tamanho_populacao=50,
    numero_geracoes=8,
    taxa_crossover=0.8,
    taxa_mutacao=0.15,
    tamanho_torneio=3,
    quantidade_elites=2,
    verbose=True
):
    """Executa a busca genetica e retorna o melhor individuo encontrado."""
    validacao = StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=SEMENTE_AG
    )
    cache_avaliacoes = {}
    populacao = criar_populacao(espaco_busca, tamanho_populacao)
    historico = []

    for geracao in range(1, numero_geracoes + 1):
        populacao_avaliada = [
            (
                individuo,
                avaliar_individuo(
                    individuo,
                    fabrica_modelo,
                    x_treino,
                    y_treino,
                    validacao,
                    cache_avaliacoes
                )
            )
            for individuo in populacao
        ]
        populacao_avaliada.sort(key=lambda item: item[1], reverse=True)

        melhor_individuo, melhor_fitness = populacao_avaliada[0]
        fitness_medio = np.mean([fitness for _, fitness in populacao_avaliada])
        historico.append({
            "geracao": geracao,
            "melhor_fitness": melhor_fitness,
            "fitness_medio": fitness_medio
        })

        if verbose:
            print(
                f"Geracao {geracao:02d} | "
                f"Melhor recall: {melhor_fitness:.4f} | "
                f"Recall medio: {fitness_medio:.4f}"
            )

        #Elitismo: preserva as melhores solucoes sem alteracao
        nova_populacao = [
            deepcopy(individuo)
            for individuo, _ in populacao_avaliada[:quantidade_elites]
        ]

        while len(nova_populacao) < tamanho_populacao:
            pai_1 = selecionar_por_torneio(populacao_avaliada, tamanho_torneio)
            pai_2 = selecionar_por_torneio(populacao_avaliada, tamanho_torneio)

            filho_1, filho_2 = aplicar_crossover(
                pai_1,
                pai_2,
                taxa_crossover
            )
            filho_1 = aplicar_mutacao(filho_1, espaco_busca, taxa_mutacao)
            filho_2 = aplicar_mutacao(filho_2, espaco_busca, taxa_mutacao)

            nova_populacao.extend([filho_1, filho_2])

        populacao = nova_populacao[:tamanho_populacao]

    #Realiza a avaliacao final porque a ultima populacao foi criada no fim do ciclo anterior
    populacao_final = [
        (
            individuo,
            avaliar_individuo(
                individuo,
                fabrica_modelo,
                x_treino,
                y_treino,
                validacao,
                cache_avaliacoes
            )
        )
        for individuo in populacao
    ]
    melhor_individuo, melhor_fitness = max(
        populacao_final,
        key=lambda item: item[1]
    )

    return deepcopy(melhor_individuo), melhor_fitness, historico
```

Experimentos com diferentes configurações do Algoritmo Genético

O Tech Challenge exige ao menos **3 experimentos** variando parâmetros do AG (população, mutação, etc.).

| Experimento | População | Gerações | Mutação | Cruzamento | Objetivo |
|-------------|-----------|----------|---------|------------|----------|
| Conservador | 8 | 10 | 0.05 | 0.70 | Baixa diversidade |
| Equilibrado | 12 | 10 | 0.15 | 0.80 | Equilíbrio exploração/explotação |
| Exploratório | 20 | 10 | 0.30 | 0.90 | Alta diversidade |

Os três experimentos comparam o recall de validação no **SVM**. Em seguida, a melhor configuração otimiza KNN, SVM e Random Forest.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))
from algoritmo_genetico import EXPERIMENTOS_AG

# --- Experimento 1, 2 e 3: comparar configs do AG no SVM ---
resultados_experimentos = []

for experimento in EXPERIMENTOS_AG:
    print(f"\n=== Experimento: {experimento['nome']} ===")
    print(experimento["descricao"])

    params = {
        k: v for k, v in experimento.items()
        if k not in {"nome", "descricao"}
    }

    melhores_parametros, melhor_recall_cv, historico = executar_algoritmo_genetico(
        fabrica_modelo=fabricas_modelos["SVM"],
        espaco_busca=espacos_busca["SVM"],
        x_treino=x_train_escalonado,
        y_treino=y_train,
        verbose=True,
        **params,
    )

    resultados_experimentos.append({
        "experimento": experimento["nome"],
        "tamanho_populacao": experimento["tamanho_populacao"],
        "numero_geracoes": experimento["numero_geracoes"],
        "taxa_mutacao": experimento["taxa_mutacao"],
        "taxa_crossover": experimento["taxa_crossover"],
        "melhor_recall_cv": melhor_recall_cv,
        "melhores_parametros": melhores_parametros,
        "historico": historico,
        "config": params,
    })

    print("Melhores hiperparametros:", melhores_parametros)
    print(f"Melhor recall CV: {melhor_recall_cv:.4f}")

comparativo_experimentos = pd.DataFrame([
    {
        "experimento": r["experimento"],
        "populacao": r["tamanho_populacao"],
        "geracoes": r["numero_geracoes"],
        "mutacao": r["taxa_mutacao"],
        "crossover": r["taxa_crossover"],
        "melhor_recall_cv": r["melhor_recall_cv"],
    }
    for r in resultados_experimentos
]).sort_values("melhor_recall_cv", ascending=False)

print("\nComparativo dos 3 experimentos:")
try:
    display(comparativo_experimentos)
except NameError:
    print(comparativo_experimentos.to_string())

melhor_experimento = max(resultados_experimentos, key=lambda item: item["melhor_recall_cv"])
print(
    f"\nConfiguração vencedora: {melhor_experimento['experimento']} "
    f"(recall CV={melhor_experimento['melhor_recall_cv']:.4f})"
)

# --- Otimização final dos 3 modelos com a config vencedora ---
configuracao_ag = {
    **melhor_experimento["config"],
    "numero_geracoes": max(melhor_experimento["config"]["numero_geracoes"], 20),
    "verbose": True,
}

resultados_ag = {}

for nome_modelo in ["KNN", "SVM", "Random Forest"]:
    print(f"\nOtimizando o modelo {nome_modelo} com config {melhor_experimento['experimento']}")

    melhores_parametros, melhor_recall_cv, historico = executar_algoritmo_genetico(
        fabrica_modelo=fabricas_modelos[nome_modelo],
        espaco_busca=espacos_busca[nome_modelo],
        x_treino=x_train_escalonado,
        y_treino=y_train,
        **configuracao_ag,
    )

    resultados_ag[nome_modelo] = {
        "melhores_parametros": melhores_parametros,
        "melhor_recall_cv": melhor_recall_cv,
        "historico": historico,
        "experimento_base": melhor_experimento["experimento"],
    }

    print("Melhores hiperparametros:", melhores_parametros)
    print(f"Melhor recall medio na validacao cruzada: {melhor_recall_cv:.4f}")
```

Treinar os modelos com os melhores hiperparametros

Após a busca, cada modelo é recriado com a melhor combinação encontrada, treinado com todo o conjunto de treinamento e utilizado para gerar as previsões do conjunto de teste. As variáveis já existentes no notebook são atualizadas para que as análises de matriz de confusão, relatório de classificação e curva ROC utilizem os modelos otimizados.


```python
#Criar e treinar os modelos otimizados
modelo_classificador_knn = criar_knn(
    resultados_ag["KNN"]["melhores_parametros"]
)
modelo_classificador_svm = criar_svm(
    resultados_ag["SVM"]["melhores_parametros"]
)
modelo_classificador_rf = criar_random_forest(
    resultados_ag["Random Forest"]["melhores_parametros"]
)

modelo_classificador_knn.fit(x_train_escalonado, y_train)
modelo_classificador_svm.fit(x_train_escalonado, y_train)
modelo_classificador_rf.fit(x_train_escalonado, y_train)

#Gerar novas predicoes para as metricas apresentadas nas proximas etapas
y_predict_knn = modelo_classificador_knn.predict(x_test_escalonado)
y_predict_svm = modelo_classificador_svm.predict(x_test_escalonado)
y_predict_rf = modelo_classificador_rf.predict(x_test_escalonado)

print("Acuracia KNN otimizado:", accuracy_score(y_test, y_predict_knn))
print("Acuracia SVM otimizado:", accuracy_score(y_test, y_predict_svm))
print("Acuracia Random Forest otimizado:", accuracy_score(y_test, y_predict_rf))
```

Comparativo baseline × modelos otimizados

Tabela com accuracy, precision, recall e F1 no conjunto de teste, antes e depois do algoritmo genético.

```python
metricas_otimizadas = {
    "KNN": calcular_metricas(y_test, y_predict_knn),
    "SVM": calcular_metricas(y_test, y_predict_svm),
    "Random Forest": calcular_metricas(y_test, y_predict_rf),
}

linhas = []
for modelo in ["KNN", "SVM", "Random Forest"]:
    base = metricas_baseline[modelo]
    otim = metricas_otimizadas[modelo]
    for metrica in ["accuracy", "precision", "recall", "f1"]:
        linhas.append({
            "modelo": modelo,
            "metrica": metrica,
            "baseline": base[metrica],
            "otimizado_ag": otim[metrica],
            "ganho": otim[metrica] - base[metrica],
        })

comparativo_baseline_vs_ag = pd.DataFrame(linhas)
try:
    display(comparativo_baseline_vs_ag.round(4))
except NameError:
    print(comparativo_baseline_vs_ag.round(4).to_string())

resumo_pivot = comparativo_baseline_vs_ag.pivot_table(
    index="modelo",
    columns="metrica",
    values=["baseline", "otimizado_ag", "ganho"],
)
try:
    display(resumo_pivot.round(4))
except NameError:
    print(resumo_pivot.round(4).to_string())
```

Utilizando a curva ROC e AUC para validar a sensibildade do modelo

```python
from sklearn import metrics
from sklearn.metrics import roc_curve, auc
```

Testando o ROC para o modelo KNN

```python
y_prob_knn = modelo_classificador_knn.predict_proba(x_test_escalonado)[:,1] 
false_positive_rate_knn, true_positive_rate_knn, thresholds_knn = roc_curve(y_test, y_prob_knn)
roc_auc_knn = auc(false_positive_rate_knn, true_positive_rate_knn)
roc_auc_knn
```
